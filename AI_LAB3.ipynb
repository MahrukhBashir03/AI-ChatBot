```python
def modify_list():
    food_list = ['Pizza', 'Burgers', 'Chips']
    print("Original List:", food_list)

    try:
        item_to_change = input("Which item would you like to change? ")
        if item_to_change not in food_list:
            raise ValueError

        new_value = input("Enter the new value: ")
        food_list[food_list.index(item_to_change)] = new_value
        print("Updated List:", food_list)

    except ValueError:
        print("Item not found in the list.")
    except Exception as e:
        print("An error occurred:", str(e))

modify_list()
```
