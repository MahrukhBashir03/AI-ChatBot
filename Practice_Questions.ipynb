```python
# Assume scores is a list of randomly-ordered numbers between 0 and 100. Write code that sorts the list, reverses the order of the items in the list, and prints the biggest value in the list.
scores = [4, 5, 1, 2, 8, 44, 56, 98, 100, 54, 64, 78, 25, 57, 97, 67, 55, 15]
scores.sort()
scores.reverse()
print("Sorted and reversed list:", scores)
print("Biggest value in the list:", max(scores))
```

```python
#Write a function that accepts a list as an argument (assume the list contains integers) and returns the total of the values in the list.
def total_list(numbers):
    total = sum(numbers)
    return total
```
