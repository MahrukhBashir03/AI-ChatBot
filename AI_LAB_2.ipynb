```python
#Functions in python
def check_number(x):
  match x: #reserved word for switch/ used for matching elements
    case 10:
      print("It is 10.")
    case 20:
      print("It is 20.")
    case _:
      print("It is not 10 or 20.")

check_number(10)
check_number(30)
```

```python
def greet(person):
  match person:
    case "A":
      print("Hello, A!")
    case "B":
      print("Hello, B!")
    case _:
      print("Hello, stranger!")

greet("A")
greet("B")
greet("C")
```

```python
def num_check(x):
  match x:
    case 10|20|30: #match 10, 20, or 30
         print(f"Matched: {x}") #f os for formatting
    case _:
      print("No match found")

num_check(10)
num_check(20)
num_check(25)
```

```python
def num_check(x):
    match x % 2 == 0 :
        case 1:
            print(f"{x} is an Even number")
        case 0:
            print(f"{x} is an Odd number")

num_check(10)
num_check(15)
num_check(22)
num_check(31)
```

```python
def reverse_list():

    numbers = input("Enter a list of numbers separated by spaces: ")

    # Split the input string into a list of strings and convert each to an integer
    num_list = list(map(int, numbers.split()))

    # Reverse the list
    reversed_list = num_list[::-1]

    # Print the reversed list
    print("Reversed list:", reversed_list)

# Call the function
reverse_list()
```

```python
def fibonacci(n):
    fib_series = [0, 1]
    for i in range(2, n):
        fib_series.append(fib_series[-1] + fib_series[-2])
    print("Fibonacci Series:", fib_series[:n])

n = int(input("Enter the number of terms for the Fibonacci series: "))
fibonacci(n)
```
